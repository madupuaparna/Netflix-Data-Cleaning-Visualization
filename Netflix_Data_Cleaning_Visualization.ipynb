```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/netflix_titles.csv')

print("Netflix dataset loaded successfully!")

print("Dataset Shape:", df.shape)

display(df.head())
```

```python

# Display the first 5 rows
print("First 5 rows:")
display(df.head())

# Check dataset size
print("Dataset Shape:", df.shape)

# Check column names
print("\nColumn Names:")
print(df.columns.tolist())

# Check data types and missing values
print("\nDataset Information:")
df.info()

# Statistical summary
print("\nStatistical Summary:")
display(df.describe(include="all"))
```

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())

print("\nMissing Values Percentage:")
print((df.isnull().sum() / len(df)) * 100)
```

```python
plt.figure(figsize=(12, 6))

sns.heatmap(df.isnull(), cbar=False, cmap='viridis')

plt.title("Missing Values Heatmap")
plt.show()
```

```python
df_cleaned = df.copy()

# Fill missing values
columns_to_fill = ['director', 'cast', 'country', 'rating']

for column in columns_to_fill:
    df_cleaned[column] = df_cleaned[column].fillna('Unknown')

# Convert date_added into datetime format
df_cleaned['date_added'] = pd.to_datetime(
    df_cleaned['date_added'].str.strip(),
    errors='coerce'
)

print("Missing values handled successfully!")

print(df_cleaned.isnull().sum())
```

```python
print("Duplicate show IDs before removal:",
      df_cleaned['show_id'].duplicated().sum())

df_cleaned = df_cleaned.drop_duplicates(subset=['show_id'])

print("Duplicate show IDs after removal:",
      df_cleaned['show_id'].duplicated().sum())

print("Dataset Shape After Removing Duplicates:",
      df_cleaned.shape)
```

```python
df_cleaned['duration_value'] = pd.to_numeric(
    df_cleaned['duration'].str.extract(r'(\d+)')[0],
    errors='coerce'
)

# Select movie durations for outlier analysis
movie_duration = df_cleaned.loc[
    df_cleaned['type'] == 'Movie', 'duration_value'
].dropna()

# Calculate IQR
Q1 = movie_duration.quantile(0.25)
Q3 = movie_duration.quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

outliers = movie_duration[
    (movie_duration < lower_limit) |
    (movie_duration > upper_limit)
]

print("Number of Outliers:", len(outliers))
```

```python
plt.figure(figsize=(10, 5))

sns.boxplot(x=movie_duration)

plt.title("Movie Duration Outlier Detection")
plt.xlabel("Duration in Minutes")

plt.show()
```

```python
df_cleaned['duration_capped'] = df_cleaned['duration_value'].copy()

movie_mask = df_cleaned['type'] == 'Movie'

df_cleaned.loc[movie_mask, 'duration_capped'] = (
    df_cleaned.loc[movie_mask, 'duration_value']
    .clip(lower=lower_limit, upper=upper_limit)
)

print("Outlier treatment completed!")
```

```python
print("Cleaned Dataset Information:")
df_cleaned.info()

print("\nRemaining Missing Values:")
print(df_cleaned.isnull().sum())

print("\nFinal Dataset Shape:")
print(df_cleaned.shape)

display(df_cleaned.head())
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df_cleaned, x='type')

plt.title("Movies vs TV Shows on Netflix")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")

plt.show()
```

```python
countries = (
    df_cleaned[df_cleaned['country'] != 'Unknown']['country']
    .str.split(',')
    .explode()
    .str.strip()
)

top_countries = countries.value_counts().head(10)

plt.figure(figsize=(10, 6))

sns.barplot(x=top_countries.values, y=top_countries.index)

plt.title("Top 10 Countries Producing Netflix Content")
plt.xlabel("Number of Titles")
plt.ylabel("Country")

plt.show()
```

```python
year_counts = df_cleaned['release_year'].value_counts().sort_index()

plt.figure(figsize=(12, 5))

plt.plot(year_counts.index, year_counts.values)

plt.title("Netflix Content Release Year Trend")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    data=df_cleaned[df_cleaned['type'] == 'Movie'],
    x='duration_value',
    bins=30,
    kde=True
)

plt.title("Distribution of Movie Durations")
plt.xlabel("Duration in Minutes")
plt.ylabel("Frequency")

plt.show()
```

```python
genres = (
    df_cleaned['listed_in']
    .dropna()
    .str.split(',')
    .explode()
    .str.strip()
)

top_genres = genres.value_counts().head(10)

plt.figure(figsize=(10, 6))

sns.barplot(x=top_genres.values, y=top_genres.index)

plt.title("Top 10 Netflix Genres")
plt.xlabel("Number of Titles")
plt.ylabel("Genre")

plt.show()
```

```python
plt.figure(figsize=(12, 5))

sns.countplot(
    data=df_cleaned,
    x='rating',
    order=df_cleaned['rating'].value_counts().index
)

plt.title("Netflix Content Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45)

plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 11))

# 1. Movies vs TV Shows
sns.countplot(
    data=df_cleaned,
    x='type',
    ax=axes[0, 0]
)
axes[0, 0].set_title("Movies vs TV Shows")

# 2. Top 10 countries
sns.barplot(
    x=top_countries.values,
    y=top_countries.index,
    ax=axes[0, 1]
)
axes[0, 1].set_title("Top 10 Countries")

# 3. Release year trend
axes[1, 0].plot(year_counts.index, year_counts.values)
axes[1, 0].set_title("Release Year Trend")
axes[1, 0].set_xlabel("Year")
axes[1, 0].set_ylabel("Number of Titles")

# 4. Top 10 genres
sns.barplot(
    x=top_genres.values,
    y=top_genres.index,
    ax=axes[1, 1]
)
axes[1, 1].set_title("Top 10 Genres")

plt.suptitle(
    "NETFLIX DATA ANALYSIS DASHBOARD",
    fontsize=18,
    fontweight='bold'
)

plt.tight_layout()

plt.savefig(
    'netflix_dashboard.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python
df_cleaned.to_csv(
    'cleaned_netflix_dataset.csv',
    index=False
)

print("Cleaned dataset saved successfully!")
```
